```python
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/drive/MyDrive/Unidad02/Airdata.csv')
```

#Ejercicio 1

```python
df['DateTime'] = pd.to_datetime(df['DateTime'],errors='coerce')
print(df.dtypes)
```

```python
df.head(3)
```

#Ejercicio 2

```python
df['Year'] = df['DateTime'].dt.year
df['Month'] = df['DateTime'].dt.month
df['Day'] = df['DateTime'].dt.day
df['Hour'] = df['DateTime'].dt.hour
df.head(3)
```

#Ejercicio 3

```python
intervalo = pd.Timedelta(days=10, hours=7, minutes=15)
df['DateTime_suma'] = df['DateTime'] + intervalo
df[['DateTime','DateTime_suma']].head(3)
```

#Ejercicio 4

```python
df['DateTime_unix'] = df['DateTime'].astype(np.int64) // 10**9
df[['DateTime','DateTime_unix']].head(3)
```

#Ejercicio 5

```python
filtrar = df['DateTime'] > '2020-11-01'
df[filtrar].head(3)
```
