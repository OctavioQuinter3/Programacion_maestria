Mostramos los datos originales

```python
# Convertir y limpieza si encuentra algo raro se hace nan
df["sale_price"] = pd.to_numeric(df["sale_price"], errors="coerce")
df["land_square_feet"] = pd.to_numeric(df["land_square_feet"], errors="coerce")
df["gross_square_feet"] = pd.to_numeric(df["gross_square_feet"], errors="coerce")

# Convertir a float
df["sale_price"] = df["sale_price"].astype(float)
df["land_square_feet"] = df["land_square_feet"].astype(float)
df["gross_square_feet"] = df["gross_square_feet"].astype(float)

# Mostrar tipos corregidos
print("Tipos de datos corregidos:")
print(df.dtypes)
```

Aqui lo que hacemos la primera parte podemos decir que es una limpieza, en el caso que econtraramos valores como ABC, $5,000, esto lo que haria seria mandarlo a NaN, y en la segunda parte ahora si convertimos, en el caso que exista NaN esto si sera aceptado para hacer float, si no hubieramos limpiado, esto hubiera causado errores porque seguiriamos con ABC o lo que sea.

Practica 16

```python
dictionary = {}

for col in df.columns:
    dictionary[col] = {
        "tipo_dato": str(df[col].dtype),
        "no_nulos": df[col].notnull().sum(),
        "valores_unicos": df[col].nunique()
    }

    # agregamos valores estadisticos en caso que sea numerico
    if pd.api.types.is_numeric_dtype(df[col]):
        dictionary[col]["min"] = df[col].min()
        dictionary[col]["max"] = df[col].max()
        dictionary[col]["media"] = df[col].mean()

print("Diccionario de datos:")
print(dictionary)
```

Realizamos con un for para cada columna el ripo de dato valores no nulos, valores unicos y si es numerico agregamos tambien los valores estadisticos de minimo, maximo y media.
