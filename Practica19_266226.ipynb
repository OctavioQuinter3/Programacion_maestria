#Ejercicio 1

```python
duplicados_duration = df.duplicated(subset=['Duration']).sum()

print("Cantidad de registros duplicados en Duration:", duplicados_duration)
```

```python
conteo_duration = df['Duration'].value_counts()

print(conteo_duration)
```

```python
df_sin_duplicados_duration = df.drop_duplicates(subset=['Duration'])

print(df_sin_duplicados_duration)
```

#Ejercicio 2

```python
duplicados_maxpulse = df.duplicated(subset=['Pulse', 'Maxpulse']).sum()

print("Cantidad de registros duplicados en PuLse y MaxPulse:", duplicados_maxpulse)
```

```python
conteo_pulse_maxpulse = df[['Pulse', 'Maxpulse']].value_counts()

print(conteo_pulse_maxpulse)
```

```python
df_pulse_last = df.drop_duplicates(subset=['Pulse','Maxpulse'], keep='last')

print(df_pulse_last)
```
